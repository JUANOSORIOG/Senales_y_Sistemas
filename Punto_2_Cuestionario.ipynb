```python
import control.matlab as control
import matplotlib.pyplot as plt
import numpy as np

# Define la frecuencia natural (valor arbitrario para demostración)
omega_n = 10

# Define diferentes relaciones de amortiguamiento
xi_small = 0.1  # Amortiguamiento muy pequeño
xi_large = 10.0 # Amortiguamiento muy grande
xi_crit = 1.0   # Amortiguamiento crítico

# Define funciones de transferencia para diferentes relaciones de amortiguamiento
# Función de transferencia de un sistema de segundo orden: G(s) = omega_n^2 / (s^2 + 2*xi*omega_n*s + omega_n^2)
G_small_xi = control.tf([omega_n**2], [1, 2*xi_small*omega_n, omega_n**2])
G_large_xi = control.tf([omega_n**2], [1, 2*xi_large*omega_n, omega_n**2])
G_crit_xi = control.tf([omega_n**2], [1, 2*xi_crit*omega_n, omega_n**2])

# Simula la respuesta al escalón
time = np.linspace(0, 5, 500) # Vector de tiempo para la simulación
response_small_xi, t_small_xi = control.step(G_small_xi, time)
response_large_xi, t_large_xi = control.step(G_large_xi, time)
response_crit_xi, t_crit_xi = control.step(G_crit_xi, time)


# Grafica las respuestas al escalón
plt.figure(figsize=(10, 6))
plt.plot(t_small_xi, response_small_xi, label=f'ξ = {xi_small} (Muy Pequeño)')
plt.plot(t_large_xi, response_large_xi, label=f'ξ = {xi_large} (Muy Grande)')
plt.plot(t_crit_xi, response_crit_xi, label=f'ξ = {xi_crit} (Amortiguado Críticamente)')

plt.title('Respuesta al Escalón de un Sistema de Segundo Orden para Diferentes Relaciones de Amortiguamiento')
plt.xlabel('Tiempo')
plt.ylabel('Respuesta')
plt.legend()
plt.grid(True)
plt.show()
```

**Para una ξ muy pequeña (0.1): El sistema es subamortiguado y la respuesta exhibirá oscilaciones significativas antes de estabilizarse.**

**Para una ξ muy grande (10.0): El sistema es sobreamortiguado y la respuesta será lenta y perezosa, tardando mucho tiempo en alcanzar el estado estable sin oscilaciones.**

**Para una ξ críticamente amortiguada (1.0): El sistema alcanza el estado estable lo más rápido posible sin ninguna oscilación o sobreimpulso.**
